# Software Design Document (SDD)
## Darkness Elements — 8 Symbols Edition
### Full Screen Bonus + Multiplier + Progressive Jackpot + Rare Collection

**Version:** 1.0
**Date:** September 2026
**Game type:** Video Slot, 5×4 grid, 40 fixed paylines

---

## 1. Overview

Darkness Elements is a 5-reel, 4-row video slot with 40 fixed paylines, set in a dark/vampire theme. The game combines four separate reward systems, each with an independent contribution to the total RTP:

| Mechanic | Type | RTP Contribution |
|---|---|---|
| Line wins (3/4/5 match) × Random Multiplier | Base game | ~92.0% |
| Progressive Jackpot (fed by 2-match) | Side reward | ~3.2% |
| Rare Collection Bonus (collect 4 rare symbols) | Bonus feature | ~0.84% |
| Full Screen Bonus (20/20 matching symbols) | Extreme jackpot | ~0.000000% (practically unreachable) |
| Gamble (double-or-nothing, 5 modes) | Side bet, fair (EV=1.0) | 0% (does not affect RTP) |

**Total simulated RTP (100,000,000 spins): 96.03%**
**Theoretical RTP (without Rare Collection): 95.22%**

*Result confirmed across two independent 100M-spin runs (95.99% and 96.03%) — consistent convergence.*

## 3. Paytable

| Symbol | 3-match | 4-match | 5-match |
|---|---|---|---|
| 🧛 / ⚰️ / 🐦‍⬛ / 🦇 (Rare) | 2.0× | 6.0× | 300.0× |
| 🧙‍♀️ / 🌙 (Medium) | 1.5× | 4.7× | 12.5× |
| 🕯️ / 🔮 (Common) | 1.0× | 1.6× | 2.0× |

Payouts are expressed in units of "× bet_per_line" (bet_per_line = total_bet / 40).

A 2-match does NOT pay directly — instead, `0.048141× bet_per_line` is added to the progressive Jackpot fund (see Section 5).

## 4. Multiplier System

On every line win (3, 4, or 5 match), a random multiplier is drawn from the following pool:

| Multiplier | Weight in pool | Theoretical probability |
|---|---|---|
| ×64 | 1 | 0.45% |
| ×32 | 2 | 0.89% |
| ×16 | 4 | 1.79% |
| ×8 | 20 | 8.93% |
| ×4 | 75 | 33.48% |
| ×2 | 122 | 54.46% |

**Average multiplier (theoretical): 4.00×**
**Average multiplier (measured, winning spins only, 100M-spin run): 3.9999×** — matches theory almost exactly.

The multiplier applies ONLY to line wins — NOT to Jackpot payouts and NOT to the Full Screen bonus. The Rare Collection bonus uses the SAME random multiplier drawn for the current spin (or ×1 if the collection is completed via a Full Screen spin).

---

## 5. Progressive Jackpot

- **Funding:** every 2-match on any of the 40 paylines adds `0.048141 × bet_per_line` to the fund.
- **Probability of a 2-match per line:** 66.49%
- **Chance of hitting per spin:** 1 / 10,000
- **Payout:** the entire accumulated fund is paid out, then the fund resets to zero.
- The jackpot check is NOT performed on a Full Screen spin (early-return mechanic).

---

## 6. Full Screen Bonus

- **Condition:** all 20 cells on screen show the same symbol.
- **Payout:** 1,000,000× bet_per_line, WITHOUT a multiplier.
- **Theoretical probability:** 1 in 1,807,112,489,733,289 spins (practically unreachable within a realistic amount of play — 0 hits in 100M tested spins, as expected).
- If the symbol belongs to the Rare group, it automatically counts toward the Rare Collection (with a ×1 multiplier).

---

## 7. Rare Collection Bonus

- **Condition:** the player must "collect" all 4 Rare symbols (🧛 ⚰️ 🐦‍⬛ 🦇), each via a separate 5-match on any payline, in any order, within the current (incomplete) collection cycle. A symbol already collected does not count again.
- **Payout on completion:** `500 × current_spin_multiplier × bet_per_line`
- **Average bonus:** 500 × 4.00 = 2,000× bet_per_line
- **Range:** from 500× (if completed on a spin with no line win / multiplier=1) to 32,000× (if completed on a spin with a ×64 multiplier)
- **Probability of a 5-match on any Rare symbol, on any line, per spin:** 0.159875% (1 in 625)
- The collection automatically resets after each completion.

---

## 8. Gamble Feature

After a winning spin (excluding Jackpot, Rare Collection, and Full Screen wins — these are excluded from gamble due to their size), the player may risk the current win on one of 5 modes:

| Mode | Chance | Payout | EV |
|---|---|---|---|
| Small (1-3) | 50% | ×2 | 1.0 (fair) |
| Big (4-6) | 50% | ×2 | 1.0 (fair) |
| Even | 50% | ×2 | 1.0 (fair) |
| Odd | 50% | ×2 | 1.0 (fair) |
| Specific number (1-6) | 16.67% | ×6 | 1.0 (fair) |

All modes are mathematically fair (0% house edge) — the win/loss depends entirely on the player's choice and luck, without changing the game's overall RTP. Unlimited consecutive double-or-nothing chains are allowed.

## 10. Volatility

| Metric | Value |
|---|---|
| Hit Frequency | 92.93% |
| Standard Deviation (σ) | 5.7173 |
| Mean (μ) | 0.9603 |
| Volatility Index (σ/μ) | 5.95 |

### 10.1 Classification: Medium-to-High (hybrid profile)

The game shows an atypical volatility profile:

- **Base game behavior (Medium volatility):** a 92.93% hit rate, with ~70% of all payouts falling in the 0–1× range — i.e., frequent small "wins," many of which don't even cover the original bet. This creates a smooth, predictable feel over short sessions.
- **Bonus feature tail (High volatility):** the Jackpot (average 315×, observed max 2,904×) and Rare Collection (2,000× average, theoretical cap 32,000×) add a rare but real heavy tail, which mathematically raises σ/μ without changing the experience of 99%+ of spins.
- The Full Screen bonus (1,000,000×) exists mathematically but is practically unreachable (1 in 1.8 quadrillion) — it does not meaningfully contribute to the perceived volatility.

**Marketing recommendation:** "Medium volatility with high-volatility bonus features," rather than simply "High volatility" — this more accurately reflects the actual player experience.

## 13. Simulation Methodology

- **Volume:** 100,000,000 independent spins, Monte Carlo method.
- **RNG source:** `numpy.random.default_rng()` (PCG64) for the simulation; the production HTML build uses a combination of `crypto.getRandomValues()` and `Math.random()`.
- **Validation:** the vectorized (numpy) implementation was verified spin-by-spin against a scalar reference implementation of the original algorithm — full agreement down to floating-point rounding (~1e-7).
- **Computation speed:** 100M spins in ~102 minutes (16,350 spins/sec); bankroll analysis an additional ~28 minutes.
- **Reproducibility:** two independent 100M-spin runs produced consistent results (Total RTP 95.99% and 96.03%; Volatility Index 5.86 and 5.95).

## 14. Known Limitations / Future Work

1. **The Full Screen bonus is statistically unreachable** within any realistic amount of play — a design decision is needed on whether it should remain a "theoretical ceiling" for marketing purposes, or be made more attainable.
2. **Rare Collection state persistence** — the mechanic retains state across multiple spins; a UI decision is needed on how this state is communicated to the player across sessions.
3. The **Gamble feature** was not included in the main RTP simulation, since it is mathematically fair (EV=1.0) — but it has not been tested for UX/psychological retention effects.
4. Regulatory review (licensing, responsible gambling mechanisms) is required before launch with real-money wagering.
